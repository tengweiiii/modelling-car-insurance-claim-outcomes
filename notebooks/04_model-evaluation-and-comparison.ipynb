## 4.2 Interpretation of the Results

**1. Driving experience achieved the highest accuracy.**

`driving_experience` achieved an accuracy of 77.71%, making it the best single feature for predicting insurance claim outcomes. This suggests that a customer's driving experience provides useful information about their likelihood of making an insurance claim.

**2. Age was another strong predictor.**

Age achieved 77.47% accuracy, only 0.24 percentage points below driving experience. This suggests that age is also strongly associated with insurance claim outcomes.

**3. Income and vehicle ownership showed relatively good performance.**

Income achieved 74.25%, while vehicle ownership achieved 73.51%. These features may provide additional information about differences in customer characteristics and claim behaviour.

**4. Several features performed at the baseline accuracy.**

Gender, education, vehicle year, marital status, children, postal code, vehicle type, speeding violations, DUIs and past accidents all achieved 68.67%. This is the same accuracy as the majority-class baseline, meaning these individual models did not improve overall accuracy over simply predicting that every customer would make no claim.

Note that this does not necessarily mean these features have no relationship with insurance claims. It means they did not improve accuracy under the current modelling approach and classification threshold.
